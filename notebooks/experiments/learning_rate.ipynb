```python
from sfs.models import make_MLP, make_CNN
from sfs.training import create_training_data, train
from sfs.config import CNNConfig
import pandas as pd
```

```python
training_data = pd.read_csv("/Users/admin/Desktop/mini_projects/digit_recognition/data/train.csv")
X = training_data.drop(columns = ["label"])
y = training_data["label"]

Data = create_training_data(X,y)

learning_rates = [10**-k for k in range(1,6)]


CNN_results = dict()
MLP_results = dict()

cnn_conf = CNNConfig()

for lr in learning_rates:
    model = make_CNN(cnn_conf)
    res = train(model, Data, learning_rate = lr)
    CNN_results[lr] = res

for lr in learning_rates:
    model = make_MLP()
    res = train(model, Data, learning_rate = lr)
    MLP_results[lr] = res



```

```python
for lr in learning_rates:
    print(lr)
    display(MLP_results[lr])
```

```python
for lr in learning_rates:
    print(lr)
    display(CNN_results[lr])
```

```python
import matplotlib.pyplot as plt

rates = sorted(learning_rates)
for metric, ylabel in [("val_accuracy", "Validation accuracy"),
                        ("val_loss", "Validation loss")]:
    fig, ax = plt.subplots(figsize=(7, 4))
    for label, results in [("CNN", CNN_results), ("MLP", MLP_results)]:
        values = [results[lr][metric].iloc[-1] for lr in rates]
        ax.plot(rates, values, marker="o", label=label)
    ax.set_xscale("log")
    ax.set_xticks(rates)
    ax.set_xticklabels([f"{lr:g}" for lr in rates])
    ax.set_xlabel("Learning rate")
    ax.set_ylabel(ylabel)
    ax.set_title(f"Final-epoch {ylabel.lower()} by learning rate")
    ax.legend()
    ax.grid(True, which="both", alpha=0.3)
    plt.show()
```
